```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
import sys

ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))

from project_4 import backpropagation as nn
```

```python
dataset_file = "rainfall_dataset.csv"
df = pd.read_csv(dataset_file)
df_norm = df.drop(df.columns[0], axis=1)

dataset_arr = df_norm.to_numpy()

print(df)
```

```python
max_data = np.max(dataset_arr)
min_data = np.min(dataset_arr)

data_norm = 0.8 * (dataset_arr - min_data) / (max_data - min_data) + 0.1
data_cols = data_norm.shape[1]

data_in = data_norm[:, :data_cols-1]
target = data_norm[:, data_cols-1:].ravel()

input_row = data_in.shape[0]
input_col = data_in.shape[1]

print("========== Data Inputs ==========")
with np.printoptions(precision=3, suppress=True):
    print(data_in)

print("\n\n============ Targets ============")
with np.printoptions(precision=3, suppress=True):
    print(target.reshape(-1, 1))
```

```python
ALPHA       = 0.9
MU          = 0.5
MAX_EPOCH   = 100000

NUM_LAYERS  = 2
NUM_NEURONS = [6, 1]
```

```python
layers = [[] for _ in range(NUM_LAYERS)]
for i in range(NUM_LAYERS):
    if i == 0:
        layers[i] = nn.Layer(NUM_NEURONS[i], input_col, i)
        layers[i].nguyen_widrow_init()

    elif i == NUM_LAYERS - 1:
        layers[i] = nn.Layer(1, NUM_NEURONS[i-1], i)

    else:
        layers[i] = nn.Layer(NUM_NEURONS[i], NUM_NEURONS[i-1], i)

    layers[i].get_info()
```

```python
epochs = 0
mse_acc = []
epochs_acc = []

stop_train = False

while True:
    epochs += 1

    error = np.zeros(input_row)
    output = np.zeros(input_row)

    for i in range(input_row):
        net_out = [[] for _ in range(NUM_LAYERS)]
        for j in range(NUM_LAYERS):
            if j == 0:
                net_out[j] = layers[j].forward_prop(data_in[i])

            else:
                net_out[j] = layers[j].forward_prop(net_out[j-1])

        error[i] = target[i] - net_out[NUM_LAYERS - 1][0]

        backprop_err = [[] for _ in range(NUM_LAYERS)]
        for j in range(NUM_LAYERS):
            rev_count = NUM_LAYERS - j

            if j == 0:
                backprop_err[j] = layers[rev_count - 1].back_prop(np.array([error[i]]), ALPHA, net_out[rev_count - 2], MU)

            elif j == NUM_LAYERS - 1:
                backprop_err[j] = layers[rev_count - 1].back_prop(backprop_err[j-1], ALPHA, data_in[i], MU)

            else:
                backprop_err[j] = layers[rev_count - 1].back_prop(backprop_err[j-1], ALPHA, net_out[rev_count - 2], MU)
                
    mse = nn.evaluate(error)
    print(f"\033[2K\rEpoch to {epochs}\n"
            f"\033[2K\rMSE\t: {mse:.7f}\n"
            f"\033[2K\rTarget\t: Inf")

    if nn.is_save(epochs):
        epochs_acc.append(epochs)
        mse_acc.append(mse)

    # if mse <= MIN_ERROR:
    #     stop_train = True

    if MAX_EPOCH and epochs >= MAX_EPOCH:
        stop_train = True

    if stop_train:
        break
    else:
        if nn.is_save(epochs):
            print("")
        else:
            print("\033[3A", end="")
```
